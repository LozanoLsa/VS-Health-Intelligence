# 01 - Data Exploration
Exploratory analysis of raw equipment and failure data.

```python
import sys
sys.path.insert(0, "..")
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option("display.max_columns", None)
```

## Load Raw Data

```python
equipment  = pd.read_csv("../data/raw/equipment_master.csv")
failures   = pd.read_csv("../data/raw/failures.csv", parse_dates=["failure_date"])
production = pd.read_csv("../data/raw/production_data.csv", parse_dates=["date"])
print(equipment.shape, failures.shape, production.shape)
```

## Equipment by VSM and Area

```python
equipment.groupby(["vsm", "area"]).size().unstack(fill_value=0)
```

## Monthly Failure Count

```python
fig, ax = plt.subplots(figsize=(12, 4))
failures.set_index("failure_date").resample("ME")["failure_id"].count().plot(
    ax=ax, color="#E84C1F", linewidth=2)
ax.set_title("Monthly Failure Count - Full Fleet")
ax.set_ylabel("Failures")
ax.set_xlabel("")
plt.tight_layout()
```

## Failure Mode Distribution

```python
failures["failure_mode"].value_counts().plot(
    kind="barh", color="#1A5276", figsize=(8, 5))
plt.title("Top Failure Modes")
plt.tight_layout()
```

## Downtime Distribution by VSM

```python
merged = failures.merge(equipment[["machine_id", "vsm"]], on="machine_id")
merged.groupby("vsm")["downtime_hrs"].describe().round(2)
```

## OEE by VSM

```python
production.groupby("vsm")["oee_pct"].describe().round(2)
```
